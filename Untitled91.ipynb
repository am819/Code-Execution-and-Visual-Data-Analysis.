```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# Load data
nodes_df = pd.read_csv('C:\\Users\\khaled\\Downloads\\got_nodes.csv')
edges_df = pd.read_csv('C:\\Users\\khaled\\Downloads\\got_edges.csv')

# Normalize house names
nodes_df['house'] = nodes_df['house'].str.strip().str.title()

# Create graph
G = nx.from_pandas_edgelist(edges_df, source="source", target="target", edge_attr="Weight")

# Add node attributes
house_dict = nodes_df.set_index("name")["house"].to_dict()
nx.set_node_attributes(G, house_dict, "house")

# Custom colors for houses
custom_colors = {
    'Stark': '#1f78b4',      # blue
    'Lannister': '#e31a1c',  # red
    'Targaryen': '#33a02c',  # green
    'Baratheon': '#ff7f00',  # orange
    'Greyjoy': '#6a3d9a',    # purple
    'Tyrell': '#b15928',     # brown
    'Martell': '#fdbf6f'     # light orange
}

# Assign colors to nodes
node_colors = [custom_colors.get(G.nodes[node]['house'], '#2ca02c') for node in G.nodes]

# Edge weights for thickness
weights = [G[u][v]['Weight'] for u, v in G.edges()]
edge_widths = [w / 20 for w in weights]  # scale down

# Draw graph
plt.figure(figsize=(18, 18))
pos = nx.spring_layout(G, k=0.3, seed=42)

# Draw nodes
nx.draw_networkx_nodes(
    G,
    pos,
    node_color=node_colors,
    node_size=1000
)

# Draw edges with weights
nx.draw_networkx_edges(
    G,
    pos,
    width=edge_widths,
    alpha=0.7
)

# Draw labels
nx.draw_networkx_labels(
    G,
    pos,
    font_size=9,
    font_weight="bold"
)

# Legend
for house, color in custom_colors.items():
    plt.scatter([], [], c=color, label=house)

plt.legend(scatterpoints=1, frameon=True, title="House")
plt.title("Game Of Thrones Season 3 Social Network Graph", fontsize=16)

# Show graph
plt.show()

# Optional: save to file
# plt.savefig("got_network.png", dpi=300, bbox_inches='tight')
```
